```python
# Step 1: Standardize column names for clarity and consistency
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
# Step 2: Check for duplicates
duplicate_rows = df.duplicated().sum()
```

```python
# Step 3: Summary of missing values per column
missing_values = df.isnull().sum()
```

```python
# Step 4: Check for anomalies (e.g., negative or zero product lengths)
negative_lengths = df[df['productlength'] < 0].shape[0]
zero_lengths = df[df['productlength'] == 0].shape[0]
```

```python
# Step 5: Summary of findings
summary = {
    "total_entries": len(df),
    "duplicate_rows": duplicate_rows,
    "missing_values": missing_values.to_dict(),
    "negative_productlength_values": negative_lengths,
    "zero_productlength_values": zero_lengths
}

summary
```

```python
# Next Step : Drop duplicate rows
df_cleaned = df.drop_duplicates()
```

```python
# Handle missing values
# - For simplicity: drop rows missing essential numeric fields
df_cleaned = df_cleaned.dropna(subset=['producttypeid', 'productlength'])
```

```python
# Step 3: Reset index for consistency
df_cleaned.reset_index(drop=True, inplace=True)
```

```python
# Step 4: Show cleaned dataset summary
cleaned_summary = {
    "new_total_entries": len(df_cleaned),
    "missing_values_post_cleaning": df_cleaned.isnull().sum().to_dict()
}

cleaned_summary
```

```python
import re
```

```python
# Function to generate short_title
def generate_short_title(title):
    # Remove common redundant words/phrases
    title = re.sub(r'\b(set of|includes|features|with|for|and|pack of|pcs|inch|cm)\b', '', title, flags=re.IGNORECASE)
    # Keep alphanumeric and basic punctuation, then strip extra spaces
    title = re.sub(r'[^A-Za-z0-9\s\'\-&:,]', '', title)
    title = re.sub(r'\s+', ' ', title).strip()
    # Truncate to 50 characters max, clean cutoff
    return title[:50].strip()
```

```python
# Apply the function to generate the short_title
df_cleaned['short_title'] = df_cleaned['title'].apply(generate_short_title)
```
